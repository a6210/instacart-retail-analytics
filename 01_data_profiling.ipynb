```python
from pathlib import Path

PRIOR_CSV = DATA_PATH / "order_products__prior.csv"
PRIOR_PARQUET_DIR = SILVER_PATH / "order_products_prior"

PRIOR_PARQUET_DIR.mkdir(parents=True, exist_ok=True)
```

```python
import pandas as pd

chunk_number = 0
total_rows = 0
CHUNK_SIZE=500000

for chunk in pd.read_csv(PRIOR_CSV, chunksize=CHUNK_SIZE):

    chunk_number += 1

    # Standardize column names
    chunk.columns = chunk.columns.str.strip().str.lower()

    # Remove exact duplicate rows
    chunk = chunk.drop_duplicates()

    # Validate reordered
    chunk = chunk[
        chunk["reordered"].isin([0, 1])
    ]

    # Validate cart position
    chunk = chunk[
        chunk["add_to_cart_order"] > 0
    ]

    # Save each chunk as a separate parquet file
    output_file = (
        PRIOR_PARQUET_DIR /
        f"part_{chunk_number:03d}.parquet"
    )

    chunk.to_parquet(
        output_file,
        index=False
    )

    total_rows += len(chunk)

    print(
        f"Chunk {chunk_number} completed | "
        f"Rows written: {len(chunk):,} | "
        f"Total: {total_rows:,}"
    )
```

NOW WE ARE READY TO CREATE GOLD LAYER

```python
GOLD_PATH=Path("data/gold")

GOLD_PATH.mkdir(parents=True,exist_ok=True)
```

```python
product_parts = []
for i, file in enumerate(analytics_files,start=1):
    df = pd.read_parquet(file,columns=[
            "product_id",
            "product_name",
            "department",
            "aisle",
            "order_id",
            "user_id",
            "reordered",
            "add_to_cart_order"
            ])

    product_summary=(df.groupby([
        "product_id",
        "product_name",
        "department",
        "aisle"
    
    ],as_index=False
    ).agg(
        total_purchases=("order_id","count"),
        unique_orders=("order_id","nunique"),
        unique_customers=("user_id","nunique"),
        reorder_count=("reordered","sum"),
        avg_cart_position=("add_to_cart_order","mean")

    )  
)    
    product_parts.append(product_summary)

    print(f"Processed file {i}/{len(analytics_files)}")         
```

```python
product_all = pd.concat(
    product_parts,
    ignore_index=True
)

print("Combined partial rows:", product_all.shape)

display(product_all.head())
```

```python
product_all["cart_position_total"] = (
    product_all["avg_cart_position"]
    *
    product_all["total_purchases"]
)
```

```python
gold_product_performance = (
    product_all
    .groupby(
        [
            "product_id",
            "product_name",
            "department",
            "aisle"
        ],
        as_index=False
    )
    .agg(
        total_purchases=("total_purchases", "sum"),
        reorder_count=("reorder_count", "sum"),
        cart_position_total=("cart_position_total", "sum")
    )
)
```

```python
gold_product_performance["reorder_rate"] = (
    gold_product_performance["reorder_count"]
    /
    gold_product_performance["total_purchases"]
)
```

```python
gold_product_performance["avg_cart_position"] = (
    gold_product_performance["cart_position_total"]
    /
    gold_product_performance["total_purchases"]
)
```

we dont need now the cart_poistion_total so we can drop it

```python
gold_product_performance.drop(
    columns=["cart_position_total"],
    inplace=True
)
```

```python
display(
    gold_product_performance
    .sort_values(
        "total_purchases",
        ascending=False
    )
    .head(20)
)
```

```python
gold_department_performance=(gold_product_performance.groupby("department",as_index=False).agg(
    total_purchases=("total_purchases","sum"),
    total_reorders=("reorder_count","sum"),
    product_count=("product_id","nunique")
)
)
```

```python
gold_department_performance["reorder_rate"]=(gold_department_performance["total_reorders"]/gold_department_performance["total_purchases"])
```

```python
gold_department_performance["purchase_share_pct"] = (
    gold_department_performance["total_purchases"]
    /
    gold_department_performance["total_purchases"].sum()
) * 100
```

now we will create the final golden table for customer metrics

```python
customer_parts=[]

for i,file in enumerate(analytics_files,start=1):
    df=pd.read_parquet(file,columns=
    [
        "user_id","order_id","product_id","reordered","days_since_prior_order"
    ])
    customer_summary=(
        df.groupby("user_id",as_index=False).agg(
            total_items=("product_id","count"),
            unique_products=("product_id","nunique"),
            reorder_count=("reordered","sum"),
            avg_days_between_order=("days_since_prior_order","mean")
        )
    )
    customer_parts.append(customer_summary)

    print(f"Processed file{i}/{len(analytics_files)}")
```

```python
customer_all = pd.concat(
    customer_parts,
    ignore_index=True
)

print(customer_all.shape)

display(customer_all.head())
```

```python
gold_customer_metrics = (
    customer_all
    .groupby(
        "user_id",
        as_index=False
    )
    .agg(
        total_items=("total_items", "sum"),
        reorder_count=("reorder_count", "sum"),
        avg_days_between_orders=("days_since_prior_order", "mean")
    )
)
```
